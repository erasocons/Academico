# Módulo de Fundamentos de Pandas: Introducción

Bienvenido a este módulo esencial para el análisis de datos en Python. Pandas es la librería principal para la manipulación y limpieza de datos estructurados.

```
import pandas as pd
import numpy as np

print(f"Versión de Pandas: {pd.__version__}")
```

## ¿Qué es una Serie y un DataFrame?

- **Series**: Un arreglo unidimensional con etiquetas (índice).
- **DataFrame**: Una estructura tabular bidimensional con filas y columnas etiquetadas.

```
# Creación de una Serie
s = pd.Series([10, 20, 30, 40], index=['a', 'b', 'c', 'd'])
print("--- Serie ---")
print(s)

# Creación de un DataFrame
data = {
    'Nombre': ['Ana', 'Luis', 'Carlos', 'Sofía'],
    'Edad': [28, 34, 25, 42],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla']
}
df = pd.DataFrame(data)
print("\n--- DataFrame ---")
print(df)
```
