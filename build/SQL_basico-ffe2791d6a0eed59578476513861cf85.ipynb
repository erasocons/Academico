# Introducción a SQL

En este cuaderno aprenderemos los fundamentos de SQL utilizando Python y SQLite para simular bases de datos relacionales.

```
import sqlite3
import pandas as pd

# Crear una base de datos en memoria
conn = sqlite3.connect(':memory:')
cursor = conn.cursor()

# Crear una tabla de ejemplo
cursor.execute('''
CREATE TABLE empleados (
    id INTEGER PRIMARY KEY,
    nombre TEXT,
    departamento TEXT,
    salario REAL,
    fecha_contratacion TEXT
);
''')

# Insertar datos de prueba
datos = [
    (1, 'Ana Pérez', 'Ventas', 3500.0, '2021-05-10'),
    (2, 'Luis Gómez', 'IT', 4500.0, '2019-03-15'),
    (3, 'María Ruiz', 'Ventas', 3200.0, '2022-01-20'),
    (4, 'Carlos López', 'IT', 5000.0, '2018-11-05'),
    (5, 'Elena Torres', 'RRHH', 3000.0, '2020-07-01')
]

cursor.executemany('INSERT INTO empleados VALUES (?, ?, ?, ?, ?);', datos)
conn.commit()
```

## SELECT, FROM y WHERE

La sentencia `SELECT` se utiliza para seleccionar datos de una base de datos. La cláusula `FROM` especifica la tabla, y `WHERE` filtra los registros.

```
query = "SELECT nombre, salario FROM empleados WHERE departamento = 'Ventas';"
df = pd.read_sql_query(query, conn)
df
```

## ORDER BY y LIMIT

`ORDER BY` se usa para ordenar el resultado en orden ascendente o descendente. `LIMIT` restringe el número de filas devueltas.

```
query = "SELECT nombre, salario FROM empleados ORDER BY salario DESC LIMIT 2;"
df = pd.read_sql_query(query, conn)
df
```

## DISTINCT

La cláusula `DISTINCT` se utiliza para devolver solo valores diferentes (únicos).

```
query = "SELECT DISTINCT departamento FROM empleados;"
df = pd.read_sql_query(query, conn)
df
```

## CASE WHEN

`CASE` es la expresión condicional de SQL, similar a un if-else en programación.

```
query = """
SELECT nombre, salario,
       CASE 
           WHEN salario > 4000 THEN 'Alto'
           ELSE 'Normal'
       END AS nivel_salarial
FROM empleados;
"""
df = pd.read_sql_query(query, conn)
df
```

## Funciones de Cadenas (String Functions)

SQL ofrece funciones para manipular textos, como `UPPER`, `LOWER` o concatenación.

```
query = "SELECT UPPER(nombre) AS nombre_mayuscula FROM empleados;"
df = pd.read_sql_query(query, conn)
df
```

## Funciones de Fecha (Date Functions)

Permiten extraer partes de una fecha o realizar operaciones con ellas.

```
query = "SELECT nombre, strftime('%Y', fecha_contratacion) AS anio FROM empleados;"
df = pd.read_sql_query(query, conn)
df
```

## Valores Nulos (NULL Values)

Manejo de registros ausentes utilizando `IS NULL` o `IS NOT NULL`.

```
cursor.execute("INSERT INTO empleados VALUES (6, 'Sofia Castro', NULL, NULL, '2023-02-01');")
conn.commit()

query = "SELECT * FROM empleados WHERE salario IS NULL;"
df = pd.read_sql_query(query, conn)
df
```
