# Introducción a la Inferencia Estadística y Pruebas de Hipótesis
En este cuaderno exploraremos los conceptos fundamentales de la inferencia estadística, centrándonos en las pruebas de hipótesis (hypothesis testing) usando Python.

```
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
np.random.seed(42)
```

## 1. Generación de Datos de Muestra
Imaginemos que estamos estudiando el efecto de un nuevo método de estudio en las calificaciones de los exámenes. Tenemos una población de control y una muestra sometida al nuevo método.

```
mu_control = 70
control_group = np.random.normal(loc=mu_control, scale=10, size=50)

mu_treatment = 74
treatment_group = np.random.normal(loc=mu_treatment, scale=10, size=50)

print(f"Media grupo control: {np.mean(control_group):.2f}")
print(f"Media grupo tratamiento: {np.mean(treatment_group):.2f}")
```

## 2. Planteamiento de Hipótesis
- **Hipótesis nula ($H_0$)**: No hay diferencia significativa entre las medias de ambos grupos ($\mu_{tratamiento} - \mu_{control} = 0$).
- **Hipótesis alternativa ($H_a$)**: El grupo de tratamiento tiene una media significativamente mayor ($\mu_{tratamiento} - \mu_{control} > 0$).

## 3. Prueba t de Student para muestras independientes
Utilizaremos la función `ttest_ind` de `scipy.stats` para realizar la prueba estadística.

```
t_stat, p_value = stats.ttest_ind(treatment_group, control_group, alternative='greater')

print(f"Estadístico T: {t_stat:.4f}")
print(f"P-valor: {p_value:.4f}")

alpha = 0.05
if p_value < alpha:
    print("Resultado: Rechazamos la hipótesis nula (H0). Hay evidencia significativa de que el tratamiento funciona.")
else:
    print("Resultado: No podemos rechazar la hipótesis nula (H0).")
```

## 4. Visualización de las Distribuciones

```
plt.figure(figsize=(10, 6))
sns.kdeplot(control_group, label='Control', fill=True, color='skyblue')
sns.kdeplot(treatment_group, label='Tratamiento', fill=True, color='orange')
plt.axvline(np.mean(control_group), color='blue', linestyle='--', label='Media Control')
plt.axvline(np.mean(treatment_group), color='darkorange', linestyle='--', label='Media Tratamiento')
plt.title('Comparación de Distribuciones entre Grupos')
plt.xlabel('Calificaciones')
plt.ylabel('Densidad')
plt.legend()
plt.show()
```
