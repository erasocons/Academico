# Fundamentos de Python

Este cuaderno cubre los conceptos esenciales de programación en Python, desde la sintaxis básica hasta el manejo de archivos y excepciones.

## 1. Sintaxis Básica, Comentarios y Variables

Python utiliza indentación para definir bloques de código en lugar de llaves. Las variables se asignan directamente con el operador `=`.

```
# Esto es un comentario de una sola línea
variable_entera = 42
variable_texto = "Hola, Data Science"
print(variable_entera, variable_texto)
```

## 2. Tipos de Datos Primitivos

Los tipos básicos incluyen enteros, flotantes, complejos, cadenas de texto, booleanos y el valor nulo (None).

```
entero = 10
flotante = 10.5
complejo = 1 + 2j
cadena = "Python"
booleano = True
nulo = None

print(type(entero), type(flotante), type(cadena), type(booleano), type(nulo))
```

## 3. Estructuras de Datos

Python ofrece colecciones integradas muy potentes: Listas, Tuplas, Diccionarios y Sets.

```
# Lista (mutable)
mi_lista = [1, 2, 3, "cuatro"]

# Tupla (inmutable)
mi_tupla = (1, 2, 3)

# Diccionario (clave-valor)
mi_diccionario = {"nombre": "Ana", "edad": 28}

# Set (elementos únicos)
mi_set = {1, 2, 2, 3, 4}
print(mi_lista, mi_tupla, mi_diccionario, mi_set)
```

## 4. Estructuras de Control (Condicionales y Bucles)

Permiten tomar decisiones y repetir bloques de código.

```
edad = 20
if edad >= 18:
    print("Mayor de edad")
elif edad > 12:
    print("Adolescente")
else:
    print("Niño")

# Bucle For
for i in range(3):
    print("Iteración for:", i)

# Bucle While con control
contador = 0
while contador < 3:
    if contador == 1:
        contador += 1
        continue
    print("Contador while:", contador)
    contador += 1
```

## 5. Funciones y Argumentos

Las funciones se definen con `def` y pueden recibir argumentos posicionales, nombrados, `*args` y `**kwargs`.

```
def sumar_todo(*args, **kwargs):
    total = sum(args)
    print("Argumentos extra:", kwargs)
    return total

resultado = sumar_todo(1, 2, 3, 4, operacion="suma")
print("Resultado:", resultado)
```

## 6. Módulos, Excepciones y Manejo de Archivos

Importación de módulos, bloques try-except-finally y lectura/escritura de archivos.

```
import math

print("Raíz cuadrada de 16:", math.sqrt(16))

try:
    resultado = 10 / 0
except ZeroDivisionError:
    print("Error: División por cero")
finally:
    print("Bloque finally ejecutado")

# Manejo de archivos (crear y leer)
with open("test.txt", "w") as f:
    f.write("Hola mundo desde un archivo en Python")

with open("test.txt", "r") as f:
    contenido = f.read()
    print("Contenido del archivo:", contenido)
```
