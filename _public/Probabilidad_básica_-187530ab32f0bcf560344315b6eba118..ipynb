## 2. Teorema de Bayes

El Teorema de Bayes es fundamental en Machine Learning (por ejemplo, en el clasificador Naive Bayes). Permite calcular la probabilidad condicional de un evento a partir de distribuciones de probabilidad a priori.

P(A|B) = [P(B|A) * P(A)] / P(B)

```
def teorema_bayes(p_a, p_b_dado_a, p_b_dado_no_a):
    p_no_a = 1 - p_a
    p_b = (p_b_dado_a * p_a) + (p_b_dado_no_a * p_no_a)
    p_a_dado_b = (p_b_dado_a * p_a) / p_b
    return p_a_dado_b

# Ejemplo: Probabilidad de tener una enfermedad dado un test positivo
p_enfermedad = 0.01
p_positivo_dado_enfermo = 0.99
p_positivo_dado_sano = 0.05

resultado = teorema_bayes(p_enfermedad, p_positivo_dado_enfermo, p_positivo_dado_sano)
print(f"Probabilidad real de tener la enfermedad dado un test positivo: {resultado:.4f}")
```

## 3. Distribuciones de Probabilidad

Las distribuciones describen cómo se reparten las probabilidades de los posibles valores de una variable aleatoria. Se dividen en discretas y continuas.

### 3.1 Distribución Normal (Gaussiana)

Es la distribución continua más importante en estadística y ciencia de datos, caracterizada por su media ($\mu$) y desviación estándar ($\sigma$).

```
mu, sigma = 0, 1
x = np.linspace(mu - 3*sigma, mu + 3*sigma, 100)
y = stats.norm.pdf(x, mu, sigma)

plt.figure(figsize=(8, 4))
plt.plot(x, y, label='Normal(0, 1)', color='blue')
plt.title('Función de Densidad de Probabilidad - Distribución Normal')
plt.xlabel('X')
plt.ylabel('Densidad de Probabilidad')
plt.legend()
plt.show()
```

### 3.2 Distribución Binomial

Distribución discreta que modela el número de éxitos en $n$ ensayos de Bernoulli independientes con probabilidad de éxito $p$.

```
n, p = 10, 0.5
k = np.arange(0, n+1)
binomial_pmf = stats.binom.pmf(k, n, p)

plt.figure(figsize=(8, 4))
plt.bar(k, binomial_pmf, color='orange', alpha=0.7)
plt.title('Distribución Binomial (n=10, p=0.5)')
plt.xlabel('Número de Éxitos (k)')
plt.ylabel('Probabilidad')
plt.show()
```

## 4. Ley de los Grandes Números (LLN)

La Ley de los Grandes Números establece que a medida que aumenta el número de ensayos en un experimento aleatorio, el promedio de los resultados obtenidos se acerca al valor esperado teórico.

```
np.random.seed(42)
tamanios_muestra = np.arange(10, 5000, 50)
medias_muestrales = [np.mean(np.random.binomial(1, 0.5, size=n)) for n in tamanios_muestra]

plt.figure(figsize=(8, 4))
plt.axhline(0.5, color='red', linestyle='--', label='Valor Teórico (0.5)')
plt.plot(tamanios_muestra, medias_muestrales, label='Media Muestral')
plt.title('Ley de los Grandes Números')
plt.xlabel('Tamaño de la Muestra')
plt.ylabel('Media')
plt.legend()
plt.show()
```
