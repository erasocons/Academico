# Guía Completa de Python: Funciones, Diccionarios y Comprehensions

En este cuaderno exploraremos tres de los conceptos más potentes y utilizados en Python:
1. **Funciones** (definición, argumentos, *args, **kwargs, lambdas)
2. **Diccionarios** (creación, métodos útiles, manipulación)
3. **Comprehensions** (List y Dict comprehensions para código más limpio y eficiente)

## 1. Funciones en Python

Las funciones nos permiten modularizar nuestro código. Veamos cómo definir una función básica y cómo usar argumentos posicionales, con nombre, y empaquetado de argumentos.

```
def calcular_precio_total(precio, impuesto=0.15, descuento=0.0):
    subtotal = precio * (1 + impuesto)
    total = subtotal * (1 - descuento)
    return round(total, 2)

print(calcular_precio_total(100))
print(calcular_precio_total(100, descuento=0.1))
```

### Argumentos variables: *args y **kwargs
- `*args` recibe un número arbitrario de argumentos posicionales como una tupla.
- `**kwargs` recibe un número arbitrario de argumentos con clave (keyword arguments) como un diccionario.

```
def reporte_usuario(nombre, *habilidades, **detalles):
    print(f"Usuario: {nombre}")
    print("Habilidades:", habilidades)
    print("Detalles adicionales:", detalles)

reporte_usuario("Ana", "Python", "SQL", "Machine Learning", edad=28, ciudad="Madrid")
```

## 2. Diccionarios en Python

Los diccionarios almacenan pares de `clave: valor`. Son mutables y altamente optimizados para búsquedas.

```
estudiante = {
    "nombre": "Carlos",
    "edad": 22,
    "cursos": ["Matemáticas", "Programación"]
}

# Acceso seguro usando .get()
print(estudiante.get("nombre"))
print(estudiante.get("promedio", "No disponible"))

# Iterar sobre claves y valores
for clave, valor in estudiante.items():
    print(f"{clave}: {valor}")
```

## 3. Comprehensions (Comprensiones)

Las comprehensions permiten construir listas, diccionarios y conjuntos de manera concisa y legible.

```
# List Comprehension: Obtener los cuadrados de los números pares del 0 al 9
cuadrados_pares = [x**2 for x in range(10) if x % 2 == 0]
print("Cuadrados pares:", cuadrados_pares)

# Dict Comprehension: Crear un diccionario que mapee números a sus cubos
cubos = {x: x**3 for x in range(5)}
print("Diccionario de cubos:", cubos)
```
