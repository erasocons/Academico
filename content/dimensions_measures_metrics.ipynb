# Dimensiones, Medidas y Métricas en Data Science

En este cuaderno exploraremos los conceptos fundamentales para el análisis de datos: 
1. **Dimensiones**: Atributos cualitativos o categóricos que describen los datos (ej. región, categoría de producto, fecha).
2. **Medidas**: Valores cuantitativos o numéricos crudos que se pueden agregar (ej. ventas, cantidad, costo).
3. **Métricas**: Fórmulas o cálculos derivados de las medidas que permiten evaluar el rendimiento (ej. margen de beneficio, tasa de conversión).

```
import pandas as pd
import numpy as np

# Crear un dataset de ejemplo para ventas e-commerce
np.random.seed(42)
n = 100

data = {
    'Fecha': pd.date_range(start='2023-01-01', periods=n, freq='D'),
    'Region': np.random.choice(['Norte', 'Sur', 'Este', 'Oeste'], n),
    'Categoria': np.random.choice(['Electrónica', 'Ropa', 'Hogar'], n),
    'Ventas_Brutas': np.random.uniform(50, 500, n).round(2),
    'Costo': np.random.uniform(30, 300, n).round(2),
    'Cantidad': np.random.randint(1, 5, n)
}

df = pd.DataFrame(data)
df.head()
```

## 1. Análisis por Dimensiones y Agregación de Medidas

Las dimensiones nos permiten segmentar nuestros datos. Vamos a agrupar las **medidas** (`Ventas_Brutas`, `Costo`) utilizando diferentes **dimensiones** como `Region` y `Categoria`.

```
# Agregando medidas por la dimensión 'Region'
resumen_region = df.groupby('Region')[['Ventas_Brutas', 'Costo', 'Cantidad']].sum()
print("Resumen por Región:")
display(resumen_region)
```

## 2. Creación de Métricas

A partir de las medidas base, podemos calcular **métricas** de negocio importantes, como el **Beneficio** y el **Margen de Beneficio (%)**.

```
# Calcular métricas derivadas
df['Beneficio'] = df['Ventas_Brutas'] - df['Costo']
df['Margen_Beneficio'] = (df['Beneficio'] / df['Ventas_Brutas']) * 100

# Mostrar las primeras filas con las nuevas métricas
df[['Region', 'Categoria', 'Ventas_Brutas', 'Costo', 'Beneficio', 'Margen_Beneficio']].head()
```

## 3. Resumen Agregado por Categoría

Ahora evaluaremos las métricas a nivel de categoría utilizando funciones de agregación.

```
# Agrupación por categoría para ver el total de ventas y el margen promedio
resumen_categoria = df.groupby('Categoria').agg(
    Ventas_Totales=('Ventas_Brutas', 'sum'),
    Costo_Total=('Costo', 'sum'),
    Beneficio_Total=('Beneficio', 'sum'),
    Margen_Promedio=('Margen_Beneficio', 'mean')
).reset()

display(resumen_categoria)
```
