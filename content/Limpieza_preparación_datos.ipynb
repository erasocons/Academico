# Limpieza y Preparación de Datos (Data Cleaning & Preparation)

En este cuaderno abordaremos las técnicas fundamentales para transformar datos crudos en datos limpios y listos para su análisis o modelado de Machine Learning.

## 1. Importación de Librerías

Primero, importamos las librerías esenciales para el manejo y manipulación de datos.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

print('Librerías importadas correctamente.')
```

## 2. Carga del Conjunto de Datos

Crearemos un DataFrame de ejemplo que simula un conjunto de datos del mundo real con problemas típicos: valores faltantes, duplicados, inconsistencias de formato y valores atípicos (outliers).

```
data = {
    'id': [1, 2, 2, 4, 5, 6, 7, 8, 9, 10],
    'nombre': ['Ana', 'Juan', 'Juan', 'Maria', 'Carlos', 'Sofia', None, 'Luis', 'Elena', 'Pedro'],
    'edad': [25, 30, 30, np.nan, 22, 45, 35, 150, 29, 31],
    'salario': ['50000', '60000$', '60000$', '45000', None, '120000', '80000', '95000', '54000', '61000'],
    'fecha_ingreso': ['2021-01-10', '2019-05-12', '2019-05-12', '2022-08-01', '2023-01-15', '2015-11-20', '2020-03-11', 'invalid_date', '2021-07-19', '2022-10-05']
}

df = pd.DataFrame(data)
df.head(10)
```

## 3. Identificación y Eliminación de Duplicados

Comprobamos si existen filas duplicadas en el conjunto de datos y las eliminamos.

```
print(f'Filas duplicadas: {df.duplicated().sum()}')
# Eliminamos duplicados basados en todas las columnas
df = df.drop_duplicates()
print(f'Dimensiones tras eliminar duplicados: {df.shape}')
```

## 4. Limpieza y Conversión de Tipos de Datos

Observamos que la columna 'salario' contiene símbolos de moneda y está en formato de texto. La columna 'fecha_ingreso' tiene un valor inválido.

```
# Limpiar la columna salario eliminando el signo '$'
df['salario'] = df['salario'].str.replace('$', '', regex=True)
df['salario'] = pd.to_numeric(df['salario'], errors='coerce')

# Convertir fecha_ingreso a datetime, forzando errores a NaT
df['fecha_ingreso'] = pd.to_datetime(df['fecha_ingreso'], errors='coerce')

df.dtypes
```

## 5. Tratamiento de Valores Faltantes (Missing Values)

Analizamos cuántos valores faltantes tenemos por columna.

```
df.isnull().sum()
```

Dependiendo del caso, podemos eliminar o imputar los valores faltantes. Para 'edad', imputaremos con la mediana. Para 'nombre', rellenaremos con 'Desconocido'.

```
mediana_edad = df['edad'].median()
df['edad'] = df['edad'].fillna(mediana_edad)

df['nombre'] = df['nombre'].fillna('Desconocido')

# Para salario, imputaremos con la media
media_salario = df['salario'].mean()
df['salario'] = df['salario'].fillna(media_salario)

df.isnull().sum()
```

## 6. Detección y Tratamiento de Outliers (Valores Atípicos)

Visualizamos la distribución de la columna 'edad' para detectar posibles anomalías (como una edad de 150 años).

```
plt.figure(figsize=(6, 4))
sns.boxplot(x=df['edad'])
plt.title('Boxplot de Edad para Detección de Outliers')
plt.show()
```

Filtramos los valores atípicos usando el rango intercuartílico (IQR) o un límite lógico (ej. edad < 100).

```
df = df[df['edad'] < 100]
df.shape
```

## 7. Normalización y Estandarización de Datos

Finalmente, preparamos las variables numéricas escalándolas para que tengan media 0 y desviación estándar 1 (Estandarización).

```
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
df[['edad_std', 'salario_std']] = scaler.fit_transform(df[['edad', 'salario']])

df.head()
```

## Conclusión

Hemos completado exitosamente las fases de:
1. Carga e inspección inicial
2. Eliminación de duplicados
3. Conversión de tipos de datos
4. Imputación de valores faltantes
5. Eliminación de outliers
6. Estandarización de variables

Los datos están ahora listos para la fase de modelado.
