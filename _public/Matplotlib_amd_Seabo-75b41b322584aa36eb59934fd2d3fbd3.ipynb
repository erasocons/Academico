# Tutorial Completo de Visualización de Datos en Python

En este cuaderno exploraremos las herramientas estándar de visualización (`Matplotlib` y `Seaborn`), analizaremos sus limitaciones en escenarios de Big Data o interactividad, y presentaremos alternativas modernas de alto rendimiento (`Plotly`, `Altair`, `Let's-Plot` y `Datashader`).

## 1. Matplotlib y Seaborn (Los Estándares)

Comenzaremos con las librerías fundamentales de Python para gráficos estáticos.

```
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np
import pandas as pd

# Datos de ejemplo
pd.set_option('display.max_columns', None)
x = np.linspace(0, 10, 100)
y = np.sin(x)

plt.figure(f'figura_mat', figsize=(8, 4))
plt.plot(x, y, label='Seno(x)', color='blue', linewidth=2)
plt.title('Gráfico Básico con Matplotlib')
plt.xlabel('Eje X')
plt.ylabel('Eje Y')
plt.legend()
plt.grid(True)
plt.show()
```

```
# Seaborn para gráficos estadísticos avanzados
df_tips = sns.load_dataset('tips')
plt.figure(figsize=(8, 5))
sns.scatterplot(data=df_tips, x='total_bill', y='tip', hue='time', style='smoker', size='size')
plt.title('Relación entre Cuenta y Propina (Seaborn)')
plt.show()
```

## 2. Problemas comunes con Matplotlib y Seaborn

Aunque son extremadamente potentes, presentan ciertos desafíos:
- **Falta de interactividad:** Los gráficos son completamente estáticos.
- **Rendimiento con Big Data:** Si intentas graficar millones de puntos, el renderizado se congela o consume toda la memoria RAM.
- **Sintaxis verbosa:** Personalizar elementos complejos en Matplotlib requiere muchas líneas de código.

## 3. Alternativas Modernas

A continuación exploraremos alternativas diseñadas para resolver estos problemas.

### A. Plotly y Plotly Express (Interactividad)
Ideal para dashboards y gráficos interactivos donde el usuario puede hacer zoom, hover y filtrar.

```
import plotly.express as px

fig = px.scatter(df_tips, x='total_bill', y='tip', color='day', size='size', hover_data=['sex'], title='Gráfico Interactivo con Plotly Express')
fig.show()
```

### B. Altair (Gramática de Gráficos Declarativa)
Basado en la especificación Vega-Lite, permite construir visualizaciones complejas mediante encadenamiento de métodos.

```
import altair as alt

chart = alt.Chart(df_tips).mark_circle(size=60).encode(
    x='total_bill',
    y='tip',
    color='day',
    tooltip=['total_bill', 'tip', 'size', 'sex']
).interactive()

chart.display()
```

### C. Let's-Plot (Estilo ggplot2 en Python)
Una librería basada en la gramática de gráficos de R (ggplot2), excelente para entornos Jupyter y con capacidad de exportación elegante.

```
from lets_plot import *

LetsPlot.setup_html()

ggplot(df_tips, aes(x='total_bill', y='tip', color='time')) + \
    geom_point(size=3) + \
    geom_smooth(method='lm') + \
    ggtitle('Gráfico estilo ggplot con Let\'s-Plot')
```

### D. Datashader (Big Data Visualization)
Cuando manejas millones de filas, las librerías tradicionales fallan. Datashader convierte los datos rasterizados en imágenes (mapas de calor de alta densidad) de manera ultra rápida.

```
import datashader as ds
import datashader.transfer_functions as tf

# Generamos un dataset grande simulado
n = 1000000
df_big = pd.DataFrame({
    'x': np.random.randn(n),
    'y': np.random.randn(n)
})

# Creamos el canvas y agregamos los puntos
cvs = ds.Canvas(plot_width=400, plot_height=400)
agg = cvs.points(df_big, 'x', 'y')
img = tf.shade(agg, cmap=['lightblue', 'darkblue'])

# Mostramos la imagen resultante
img
```
