# Flujo de Trabajo del Analista de Datos (Data Analyst Workflow)

Este cuaderno interactivo demuestra un flujo de trabajo completo para un proyecto de análisis de datos utilizando Python. Las etapas cubiertas son:
1. **Carga y Generación de Datos**: Simulación de un conjunto de datos de ventas.
2. **Limpieza y Preparación de Datos**: Tratamiento de nulos y tipos de datos.
3. **Análisis Exploratorio de Datos (EDA)**: Estadísticas descriptivas y agregaciones.
4. **Visualización de Datos**: Gráficos clave utilizando Matplotlib y Seaborn.
5. **Conclusiones**: Resumen de los hallazgos principales.

No hay una fase de modelado avanzado (como Machine Learning o analítica predictiva) después de la visualización porque ese tipo de modelado no es la función principal de un analista de datos.Si un analista de datos realiza un modelo básico, este se hace antes o durante la fase de visualización, nunca después.Aquí te explico detalladamente por qué la estructura del flujo de trabajo funciona así:

🙅 Separación de Roles (Analista vs. Científico de Datos)

1. 📊El Analista de Datos: Mira el pasado y el presente. Su objetivo es entender qué pasó, por qué pasó y comunicar esos hallazgos mediante reportes y gráficos [Dashboard]. Su flujo de trabajo termina cuando la empresa entiende la situación actual y toma una decisión.
 
 2. 🤖 El Científico de Datos: Mira el futuro. Si el negocio decide que, tras ver los gráficos del analista, necesitan un sistema que prediga automáticamente las ventas del próximo año, el proyecto pasa a manos de un Científico de Datos (Data Scientist). Él es quien se encarga de la fase de modelado predictivo complejo y entrenamiento de algoritmos

## 1. Importación de Librerías

Primero, importamos las librerías estándar necesarias para la manipulación y visualización de datos.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo para las gráficas
sns.set_theme(style="whitegrid")
%matplotlib inline
```

## 2. Generación y Carga del Dataset

Para este ejemplo, vamos a simular un dataset de ventas de una tienda en línea que contiene información sobre fecha, categoría de producto, cantidad vendida, precio unitario y región.

```python
# Semilla para reproducibilidad
np.random.seed(42)

# Fechas de un año
fechas = pd.date_range(start="2023-01-01", end="2023-12-31", freq="D")

# Datos sintéticos
n_filas = 500
data = {
    "Fecha": np.random.choice(fechas, n_filas),
    "Categoria": np.random.choice(["Electrónica", "Ropa", "Hogar", "Deportes"], n_filas),
    "Cantidad": np.random.randint(1, 10, n_filas),
    "Precio_Unitario": np.round(np.random.uniform(10.0, 500.0, n_filas), 2),
    "Region": np.random.choice(["Norte", "Sur", "Este", "Oeste"], n_filas)
}

df = pd.DataFrame(data)

# Introducimos algunos valores nulos artificialmente para practicar la limpieza
df.loc[df["Cantidad"].sample(10, random_state=42).index, "Cantidad"] = np.nan

# Visualizamos las primeras filas
df.head()
```

## 3. Limpieza y Preparación de Datos

Revisamos la estructura del DataFrame, identificamos valores faltantes y aplicamos las transformaciones necesarias.

```python
# Información general del DataFrame
print(df.info())
```

```python
# Rellenamos los valores nulos de 'Cantidad' con la mediana de la columna
mediana_cantidad = df["Cantidad"].median()
df["Cantidad"].fillna(mediana_cantidad, inplace=True)

# Creamos una nueva columna calculada: 'Venta_Total'
df["Venta_Total"] = df["Cantidad"] * df["Precio_Unitario"]

# Verificamos que no queden nulos
print(df.isnull().sum())
```

## 4. Análisis Exploratorio de Datos (EDA)

Calculamos estadísticas descriptivas básicas para entender la distribución de las variables numéricas y categóricas.

```python
# Estadísticas descriptivas
df.describe()
```

```python
# Ventas totales por categoría
ventas_por_categoria = df.groupby("Categoria")["Venta_Total"].sum().reset_index()
print(ventas_por_categoria)
```

## 5. Visualización de Datos

Generamos gráficos para comunicar visualmente los resultados obtenidos del análisis.

```python
# Gráfico de barras: Ventas Totales por Categoría
plt.figure(figsize=(8, 5))
sns.barplot(data=ventas_por_categoria, x="Categoria", y="Venta_Total", palette="Blues_d")
plt.title("Ventas Totales por Categoría de Producto")
plt.xlabel("Categoría")
plt.ylabel("Venta Total ($)")
plt.show()
```

```python
# Gráfico de cajas (Boxplot): Distribución de precios por región
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, x="Region", y="Precio_Unitario", palette="Set2")
plt.title("Distribución de Precios Unitarios por Región")
plt.xlabel("Región")
plt.ylabel("Precio Unitario ($)")
plt.show()
```

**El analista de datos no solo dice qué pasó (eso es solo reportar), sino que investiga por qué pasó.** 

En el mundo profesional, esta función se conoce técnicamente como Analítica Diagnóstica.🔍 ¿Cómo encuentra el analista las causas de algo?Para entenderlo de forma sencilla, el analista opera como un detective de datos a través de estos pasos:

1.   (El síntoma): El negocio nota un problema. Ejemplo: "Las ventas de la app cayeron un 20% este mes".
2. (La investigación): El analista cruza variables, filtra fechas y compara segmentos.Paso
3. (La causa raíz): El analista encuentra el motivo real. Ejemplo: "La caída ocurrió solo en usuarios de Android después de la última actualización del día 5".

## 6. Conclusiones

- El proceso de limpieza permitió corregir los valores faltantes sin perder registros valiosos.
- La categoría **Electrónica** (o la categoría líder resultante) muestra un desempeño destacado en volumen de ventas.
- Las regiones presentan distribuciones de precios unitarios estables, lo que indica una política de precios homogénea.
