# Tutorial: Exploración de Patrones y Relaciones en Datos

En este cuaderno aprenderemos las técnicas fundamentales para descubrir patrones, tendencias y relaciones estadísticas entre variables utilizando Python.

## 1. Importación de Librerías

Primero, importamos las librerías necesarias para el análisis de datos y la visualización.

```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo
sns.set_theme(style='whitegrid')
%matplotlib inline
```

## 2. Carga del Conjunto de Datos

Para este tutorial utilizaremos un conjunto de datos clásico de seaborn: 'diamonds', que contiene información sobre las características de los diamantes y sus precios.

```
df = sns.load_dataset('diamonds')
df.head()
```

## 3. Análisis Exploratorio Inicial (EDA)

Revisemos los tipos de datos y los valores nulos para entender la estructura general.

```
print(df.info())
print(df.describe())
```

## 4. Exploración de Relaciones Bivariadas

Utilizamos diagramas de dispersión (scatter plots) para examinar la relación entre variables numéricas, como el peso (`carat`) y el precio (`price`).

```
plt.figure(figsize=(10, 6))
sns.scatterplot(data=df.sample(1000), x='carat', y='price', hue='cut', alpha=0.7)
plt.title('Relación entre Carat, Price y Cut')
plt.show()
```

## 5. Matriz de Correlación

Para detectar relaciones lineales fuertes entre todas las variables numéricas, calculamos la matriz de correlación de Pearson y la visualizamos con un mapa de calor (heatmap).

```
plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=[np.number])
corr = numeric_df.corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Mapa de Calor de Correlaciones')
plt.show()
```

## 6. Conclusiones

Hemos explorado cómo identificar patrones visuales y estadísticos en los datos. La combinación de gráficos bivariados y matrices de correlación es clave para cualquier proyecto de Machine Learning.
