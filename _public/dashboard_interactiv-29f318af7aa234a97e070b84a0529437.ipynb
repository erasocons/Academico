## 2. Importación de Librerías

Importamos Dash, los componentes HTML y Core de Dash, Plotly Express para la generación rápida de gráficos y Pandas para la manipulación de datos.

```
import dash
from dash import dcc, html
from dash.dependencies import Input, Output
import plotly.express as px
import pandas as pd
```

## 3. Preparación de Datos

Para este ejemplo, utilizaremos un conjunto de datos integrado en Plotly Express (Gapminder) para visualizar la relación entre el PIB per cápita y la esperanza de vida.

```
# Cargar datos de ejemplo
df = px.data.gapminder()
print(df.head())
```

## 4. Estructura y Layout del Dashboard

El layout define la apariencia visual del dashboard. Utilizaremos componentes HTML para los títulos y controles deslizantes (sliders), y un componente `dcc.Graph` para mostrar el gráfico interactivo.

```
app = dash.Dash(__name__)

app.layout = html.Div([
    html.H1("Dashboard Interactivo de Población y Economía", style={'textAlign': 'center'}),
    
    html.Div([
        html.Label("Selecciona el año:"),
        dcc.Slider(
            id='year-slider',
            min=df['year'].min(),
            max=df['year'].max(),
            step=5,
            value=df['year'].min(),
            marks={str(year): str(year) for year in df['year'].unique()}
        )
    ], style={'width': '80%', 'padding': '20px', 'margin': 'auto'}),
    
    dcc.Graph(id='indicator-graphic')
])
```

## 5. Interactividad con Callbacks

Los callbacks conectan los componentes de entrada (como el slider) con los componentes de salida (como el gráfico), permitiendo que el dashboard responda dinámicamente a las acciones del usuario.

```
@app.callback(
    Output('indicator-graphic', 'figure'),
    [Input('year-slider', 'value')]
):
def update_graph(selected_year):
    filtered_df = df[df.year == selected_year]
    
    fig = px.scatter(
        filtered_df,
        x="gdpPercap",
        y="lifeExp",
        size="pop",
        color="continent",
        hover_name="country",
        log_x=True,
        size_max=55,
        title=f"Año: {selected_year}"
    )
    
    fig.update_layout(transition_duration=500)
    return fig
```

## 6. Ejecución de la Aplicación

Finalmente, ejecutamos el servidor de la aplicación Dash. (Nota: Si estás ejecutando esto en un entorno tipo Jupyter Notebook, puedes usar `app.run_server(mode='inline')` o simplemente ejecutarlo de manera estándar).

```
if __name__ == '__main__':
    app.run_server(debug=True, port=8050)
```
