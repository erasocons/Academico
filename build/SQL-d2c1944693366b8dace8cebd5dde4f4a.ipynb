## 1. ¿Qué es SQL?

**SQL** (Structured Query Language, por sus siglas en inglés) es un lenguaje estándar de programación diseñado para administrar, consultar y manipular bases de datos relacionales.

Las bases de datos relacionales almacenan información en tablas compuestas por filas (registros) y columnas (campos).

## 2. Configuración del entorno en Python

Python incluye una librería nativa llamada `sqlite3` que nos permite interactuar con bases de datos SQLite sin necesidad de instalar servidores externos. También utilizaremos `pandas` para estructurar mejor nuestros datos.

```
import sqlite3
import pandas as pd

print('Librerías importadas correctamente.')
```

## 3. Creación de una base de datos y una tabla

Vamos a crear una base de datos en memoria (o en un archivo temporal) y una tabla llamada `empleados`.

```
# Conectarse a una base de datos local (creará un archivo llamado 'empresa.db')
conexion = sqlite3.connect('empresa.db')
cursor = conexion.cursor()

# Crear la tabla empleados
cursor.execute('''
CREATE TABLE IF NOT EXISTS empleados (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    nombre TEXT NOT NULL,
    departamento TEXT NOT NULL,
    salario REAL
)
''')

# Confirmar los cambios
conexion.commit()
```

## 4. Insertar datos en la tabla

Añadiremos algunos registros de ejemplo para poder realizar consultas posteriormente.

```
# Insertar datos de prueba
datos = [
    ('Ana Pérez', 'Ventas', 3000),
    ('Carlos Gómez', 'IT', 4500),
    ('Elena Torres', 'Marketing', 3200),
    ('David Ruiz', 'IT', 4800),
    ('Sofía Castro', 'Ventas', 3100)
]

cursor.executemany('INSERT INTO empleados (nombre, departamento, salario) VALUES (?, ?, ?)', datos)
conexion.commit()

print('Datos insertados exitosamente.')
```

## 5. Consultar SQL desde Python con Pandas

Una de las ventajas más grandes en Data Science es usar `pd.read_sql_query` para transformar directamente el resultado de una consulta SQL en un DataFrame de Pandas.

```
# Consulta para traer todos los empleados
query = 'SELECT * FROM empleados'
df_empleados = pd.read_sql_query(query, conexion)

df_empleados
```

### Consultas más avanzadas

Podemos filtrar información, por ejemplo, seleccionando únicamente a los empleados del departamento de IT.

```
query_it = "SELECT nombre, salario FROM empleados WHERE departamento = 'IT'"
df_it = pd.read_sql_query(query_it, conexion)

df_it
```

También podemos realizar agregaciones, como calcular el salario promedio por departamento.

```
query_avg = '''
SELECT departamento, AVG(salario) as salario_promedio 
FROM empleados 
GROUP BY departamento
'''
df_avg = pd.read_sql_query(query_avg, conexion)

df_avg
```

## 6. Cierre de la conexión

Es una buena práctica cerrar la conexión a la base de datos cuando terminemos de trabajar con ella.

```
conexion.close()
print('Conexión cerrada correctamente.')
```
