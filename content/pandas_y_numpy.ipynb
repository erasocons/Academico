# Manejo Moderno de Datos: Pandas, NumPy y sus Alternativas

Durante años, **Pandas** y **NumPy** han sido los pilares fundamentales del ecosistema de Ciencia de Datos en Python. Sin embargo, el crecimiento exponencial de los datos ha expuesto ciertas limitaciones o cuellos de botella en estas herramientas tradicionales. En este cuaderno exploraremos estas limitaciones y conoceremos las nuevas alternativas modernas.

```
import numpy as np
import pandas as pd

# Ejemplo básico con NumPy y Pandas
array_np = np.array([1, 2, 3, 4, 5])
df_pd = pd.DataFrame({'A': array_np, 'B': array_np * 2})
print(df_pd)
```

## Cuellos de Botella de Pandas y NumPy

### Cuellos de botella en Pandas:
1. **Ejecución en un solo núcleo (Single-threaded):** Por defecto, Pandas procesa las operaciones usando un solo núcleo de la CPU, desaprovechando los procesadores multicore modernos.
2. **Alto consumo de memoria RAM:** Las estructuras internas de Pandas (como DataFrames con múltiples tipos de datos mixtos o índices complejos) consumen mucha más memoria que el tamaño real del dataset.
3. **Sintaxis ambigua:** Múltiples formas de hacer la misma operación (ej. filtrado con loc, iloc, o indexación booleana directa) que pueden generar confusión y advertencias como el temido `SettingWithCopyWarning`.

### Cuellos de botella en NumPy:
1. **Falta de soporte nativo para GPU:** NumPy está optimizado exclusivamente para CPU. No puede acelerar operaciones matriciales pesadas utilizando tarjetas gráficas.
2. **Ausencia de cálculo de gradientes automáticos:** Aunque es excelente para operaciones lineales, NumPy no calcula derivadas automáticas, algo crucial para entrenar modelos de Deep Learning.

## Nuevas Opciones para Reemplazar o Complementar Pandas

Para solucionar los problemas de rendimiento de Pandas, han surgido alternativas muy potentes:
- **Polars:** Escrita en Rust, utiliza ejecución paralela en múltiples núcleos, evaluación perezosa (lazy evaluation) y un uso optimizado de la memoria.
- **DuckDB:** Una base de datos analítica embebida (OLAP) que permite ejecutar consultas SQL ultra rápidas directamente sobre archivos CSV, Parquet o DataFrames de Pandas.

```
# Demostración conceptual de Pandas vs Polars vs DuckDB
import polars as pl
import duckdb

# Polars dataframe
df_pl = pl.DataFrame({'A': [1, 2, 3, 4, 5], 'B': [10, 20, 30, 40, 50]})
print('Polars DataFrame:')
print(df_pl)

# DuckDB query sobre Pandas
result_duckdb = duckdb.query('SELECT AVG(A), SUM(B) FROM df_pd').fetchall()
print('\nDuckDB Agregación sobre Pandas:')
print(result_duckdb)
```

## Nuevas Opciones para Reemplazar o Complementar NumPy

Para superar las limitaciones de NumPy en GPU y cálculo avanzado, hoy contamos con:
- **CuPy:** Una librería con la misma API que NumPy pero ejecutada directamente en GPUs NVIDIA, logrando aceleraciones masivas.
- **PyTorch / JAX:** Librerías orientadas a Machine Learning y Deep Learning que ofrecen tensores ejecutados en GPU/TPU y **diferenciación automática** (gradientes).

```
import torch

# Ejemplo de tensor en PyTorch con soporte de gradientes
x = torch.tensor([2.0, 3.0], requires_grad=True)
y = x * 2 + 5
z = y.sum()

# Cálculo automático de gradientes (Backpropagation)
z.backward()
print('Gradientes (dy/dx):', x.grad)
```

## ¿Cuál deberías elegir según tu caso?

- **Usa Pandas y NumPy si:** Estás trabajando con datasets pequeños (menos de unos pocos gigabytes), prototipando rápidamente, o utilizando librerías tradicionales de Scikit-Learn que dependen de ellos.
- **Usa Polars si:** Manejas datasets medianos o grandes (varios gigabytes) en una sola máquina y necesitas procesamiento paralelo rápido y sintaxis clara.
- **Usa DuckDB si:** Prefieres trabajar con consultas SQL complejas sobre grandes volúmenes de datos tabulares almacenados en archivos locales.
- **Usa CuPy si:** Tienes operaciones matriciales masivas que NumPy ejecuta lento y dispones de una tarjeta gráfica NVIDIA compatible.
- **Usa PyTorch o JAX si:** Estás construyendo modelos de Machine Learning avanzados o redes neuronales que requieren cálculo de gradientes y aceleración por hardware.
